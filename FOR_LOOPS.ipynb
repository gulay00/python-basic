```
# 1) 0-dan 4-ə qədər '#' çap edir
for x in range(5):
    print(x * "#")
```

```
# 2) Forma bənzər naxış çap edir
a = int(input("Qiyməti daxil et: "))
for x in range(a):
    print("*"*5 + " " * (a-x) + "*" * (2*x + 1) + " " * (a-x) + "*"*5)
```

```
# 3) Verilmiş mətnin hər bir simvolunun sayını tapır
list1 = "Lorem Ipsum is simply dummy text of the printing"
dict = {}
for i in set(list1):
    dict[i] = list1.count(i)
print(dict)
```

```
# 4) Dictionary-dəki dəyərlərin cəmini tapır
dict = {"a":5, "b":25, "c":17, "d":5}
sum = 0
for x in dict:
    sum = sum + dict[x]
print("Cem:", sum)
```

```
# 5) Meyvələri bir-bir çap edir
fruits = ["alma", "armud", "banan"]
for x in fruits:
    print(x)
```

```
# 6) Tuple siyahısından adları və yaşları ayırır
people = [("Ali", 25), ("Veli", 30), ("Aysel", 22)]
names = []
ages = []
for x in people:
    names.append(x[0])
    ages.append(x[1])
print(names)
print(ages)
```

```
# 7) Sözləri və uzunluqlarını çap edir
words = ["alma", "armud", "banan"]
for x in words:
    print(x, len(x))
```

```
# 8) Listdəki ədədlərin kvadratını hesablayır
numbers = [2, 3, 4]
squared = []
for x in numbers:
    squared.append(x**2)
print(squared)
```

```
# 9) Cümlədəki 'a' hərflərinin sayını tapır
sentence = "Java and JavaScript are not the same"
a_count = 0
for x in sentence:
    if x == "a":
        a_count += 1
print(a_count)
```

```
# 10) 1-dən 10-a qədər kvadrat siyahısı
squared = [x**2 for x in range(1, 11)]
print(squared)
```

```
# 11) Tələbələrin yaşlarını toplayır
summ = 0
students = [["Ali", 20], ["Veli", 15], ["Aysel", 17]]
for x in students:
    summ += x[1]
print(summ)
```

```
# 12) Siyahıdakı təkrarlanan elementləri tapır
l = [5, 5, 6, 88, 88, 88, 9]
n2 = []
def again(l):
    for x in l:
        if l.count(x) > 1:
            n2.append(x)
    return set(n2)
print(again(l))
```

```
# 13) Kvadrat kökü tam çıxan ədədləri tapır
nums = [4, 9, 16, 25, 1525]
sqr = []
for x in nums:
    for i in range(x):
        if i*i == x:
            sqr.append(i)
print(sqr)
```

```
# 14) 3-ə bölünən ədədlərin 2 qatını tapır
nums = [1, 4, 7, 12, 15, 18]
filtered = []
for n in nums:
    if n % 3 == 0:
        filtered.append(n * 2)
print(filtered)
```

```
# 15) Hər ədədi 1 azaldıb cəmləyir
nums = [3, 1, 4, 1, 5, 9]
result = 0
for n in nums:
    result += (n - 1)
print(result)
```

```
# 16) Textdə tək indexləri götürmədən string yığır
text = "programming"
text_l = ""
for x in range(len(text)):
    if x % 2 != 1:
        text_l += text[x]
print(text_l)
```

```
# 17) Sözlərin ilk və son hərflərini birləşdirir
sentence = "python is very cool"
sen = sentence.split()
res = ""
for x in sen:
    res += x[0] + x[-1] + " "
print(res)
```

```
# 18) Minimum və maksimum ədədləri listdən silir
nums = [10, 3, 7, 3, 15, 10, 15, 4]
for x in nums:
    if x == min(nums):
        nums.remove(x)
    if x == max(nums):
        nums.remove(x)
print(sorted(nums))
```

```
# 19) Kvadrat kökü integer olanları tapır
nums = [4, 9, 16, 25, 36]
sqr = []
for x in nums:
    for i in range(x):
        if i*i == x:
            sqr.append(i)
print(sqr)
```

```
# 20) Cüt ədədləri 2 qat, tək ədədləri yarıya bölür
nums = [6, 7, 9, 10, 12, 14]
nums2 = []
for x in range(len(nums)):
    if nums[x] % 2 == 0:
        nums2.append(nums[x] * 2)
    else:
        nums2.append(nums[x] // 2)
print(nums2)
```
