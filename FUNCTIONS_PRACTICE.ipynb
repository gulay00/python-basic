```
l= [5,5,6,88,88,88,9]
n2=[]
def again(l):
    for x in l:
        if l.count(x) > 1:
            n2.append(x)
    return set(n2)
print(again(l))
```

```
def square(n):
    return n * n

print(square(5))
```

```
def is_even(n):
    return n % 2 == 0

print(is_even(4), is_even(7))
```

```
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

print(factorial(5))
```

```
def reverse_string(s):
    return s[::-1]

print(reverse_string("Gulay"))
```

```
def count_vowels(s):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in s:
        if ch in vowels:
            count += 1
    return count

print(count_vowels("Data Science"))
```

```
def max_in_list(lst):
    max_val = lst[0]
    for x in lst:
        if x > max_val:
            max_val = x
    return max_val

print(max_in_list([3, 7, 2, 9, 5]))
```

```
def unique_elements(lst):
    result = []
    for x in lst:
        if x not in result:
            result.append(x)
    return result

print(unique_elements([1,2,2,3,3,3,4]))
```

```
def fibonacci(n):
    seq = []
    a, b = 0, 1
    while len(seq) < n:
        seq.append(a)
        a, b = b, a + b
    return seq

print(fibonacci(10))
```

```
def sum_of_digits(n):
    s = 0
    for ch in str(n):
        s += int(ch)
    return s

print(sum_of_digits(12345))
```

```
def is_palindrome(s):
    s = str(s)
    return s == s[::-1]

print(is_palindrome("level"), is_palindrome("python"))
```
